```python
#Creating a dataset
import pandas as pd
dict4={
    "Name":["Rohit","Pritam","Anushka","Pooja","Mohit"],
    "Department":["CSE","ECE","CSE","CSE","IT"],
    "Marks":[88,94,84,76,98],
    "City":["Delhi","Mumbai","Kolkata","Lucknow","Chennai"]
}
df=pd.DataFrame(dict4)
df.to_csv('student_data3.csv',index=False)
print(df)
```

```python
#Calculating total marks from the dataset
import pandas as pd
df=pd.read_csv('student_data3.csv')
df=df['Marks'].sum()
print("Total Marks:",df)
```

```python
#Calculating average marks from the dataset
import pandas as pd
df=pd.read_csv('student_data3.csv')
df=df['Marks'].mean()
print("Average Marks:",df)
```

```python
#Calculating maximum marks from the dataset
import pandas as pd
df=pd.read_csv('student_data3.csv')
df=df['Marks'].max()
print("Maximum Marks:",df)
```

```python
#Calculating minimum marks from the dataset
import pandas as pd
df=pd.read_csv('student_data3.csv')
df=df['Marks'].min()
print("Minimum Marks:",df)
```
