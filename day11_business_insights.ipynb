```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('data.csv')
df1=df['Marks'].sum()
df2=df['Marks'].mean()
df3=df['Marks'].max()
df4=df['Marks'].min()
df5=df['Marks'].count()

print("Sum of Marks:",df1)
print("Average Marks:",df2)
print("Maximum Marks:",df3)
print("Minimum Marks:",df4)
print("Number of Students:",df5)

x=df['Name']
y=df['Marks']
plt.title('Pie Chart for Student Marks')
plt.pie(y,labels=x,autopct='%0.01f%%')
plt.show()
```

Business Insights:-

1. Overall Performance: The dataset presents that the average marks of all the students is 87.16, which indicates that the overall performance of the class is good.

2.  Performance Range: The highest marks obtained is 98 and the lowest marks obtained is 76. The gap between them is 22 marks. This shows that performance of students is balanced and some student may require a little more attention.

3. Top Scoring Department: Students in the CSE department scored the most marks (85+) in the evaluation.

4. No. of Students: The dataset contains 5 students, which were used for analysis.

5. View through Chart: The pie chart distribution shows that high scorers cover more than half almost 53.5% of the total score.
