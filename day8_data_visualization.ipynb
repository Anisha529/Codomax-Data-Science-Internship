```python
#Creating a dataset
import pandas as pd
dict={
    "Name":["Rohini","Pritam","Anushka","Poonam","Mohit"],
    "Marks":[98,94,84,76,88],
}
df=pd.DataFrame(dict)
df.to_csv('student_marks.csv',index=False)
print(df)
```

```python
#Bar Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('student_marks.csv')
x=df['Name']
y=df['Marks']
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Bar Chart for Student Marks')
plt.bar(x,y)
plt.show()
```

```python
#Line Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('student_marks.csv')
x=df['Name']
y=df['Marks']
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Line Chart for Student Marks')
plt.plot(x,y)
plt.show()
```

```python
#Pie Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('student_marks.csv')
x=df['Name']
y=df['Marks']
plt.title('Pie Chart for Student Marks')
plt.pie(y,labels=x,autopct='%0.01f%%')
plt.show()
```
