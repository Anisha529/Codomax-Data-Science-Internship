```python
#Operators
a = 15
b = 5

# Arithmetic Operators
print("Addition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)

# Comparison Operators
print("a>b:",a>b)
print("a==b:",a==b)

#Logical Operator
print("(a>b) and (b>0):",(a>b) and (b>0))
print("(a<b) or (b>0):",(a<b) and (b>0))
```

```python
#Loops
#While Loop
i=1
sum=0
while i<=7:
    sum+=i
    i+=1
print("The sum is:",sum)
```
