```python
#Creating a CSV dataset with missing values
import pandas as pd
import numpy as np
dict2={
    "Name":["Rahul","Priya","Rohan","Anushka",np.nan,"Amit"],
    "Age":[np.nan,21,23,22,21,20],
    "Department":["CSE","ECE",np.nan ,"CSE","CSE","IT"],
    "Marks":[85,96,74,87,np.nan,90],
    "City":["Delhi",np.nan,"Kolkata","Lucknow","Chennai","Hyderabad"]
}
df=pd.DataFrame(dict2)
df.to_csv('student_data.csv',index=False)
print(df)
```

```python
#Checking for missing values
import pandas as pd
df=pd.read_csv('student_data.csv')
print(df.isnull().sum())
```

```python
#Dropping important rows with missing values
import pandas as pd
df=pd.read_csv('student_data.csv')
df=df.dropna(subset=["Name","Age"])
print(df)
```

```python
#Filling missing values with default values
import pandas as pd
df=pd.read_csv('student_data.csv')
df=df.fillna({"Department":"Unknown","City":"Unknown"})
print(df)
```

```python
#Correcting data types
import pandas as pd
df=pd.read_csv('student_data.csv')
df["Age"]=df["Age"].astype(bool)
print(df)
```

```python
#Added duplicate records to the dataset
import pandas as pd
df=pd.read_csv('student_data.csv')
print(df)
```

```python
#Dropping duplicate records
import pandas as pd
df=pd.read_csv('student_data.csv')
df=df.drop_duplicates()
print(df)
```
