```python
#Creating a dataset
import pandas as pd
dict3={
    "Name":["Riya","Pritam","Anushka","Praneet","Mohit"],
    "Department":["CSE","ECE","CSE","CSE","IT"],
    "Age":[18,21,20,22,21],
    "Marks":[85,96,74,87,92],
    "City":["Delhi","Mumbai","Kolkata","Lucknow","Chennai"]
}
df=pd.DataFrame(dict3)
df.to_csv('student_data2.csv',index=False)
print(df)
```

```python
#Filtering rows
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[df['Marks']>80]
print(df)
```

```python
#Filtering columns
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[['Department','City']]
print(df)
```

```python
#Sorting the dataset in ascending order based on Age
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[df['Age']<25].sort_values(by='Age',ascending=True)
print(df)
```

```python
#Sorting the dataset in descending order based on Age
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[df['Age']<25].sort_values(by='Age',ascending=False)
print(df)
```

```python
#Sorting the dataset in ascending order based on Marks
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[df['Marks']>70].sort_values(by='Marks',ascending=True)
print(df)
```

```python
#Sorting the dataset in descending order based on Marks
import pandas as pd
df=pd.read_csv('student_data2.csv')
df=df[df['Marks']>70].sort_values(by='Marks',ascending=False)
print(df)
```
