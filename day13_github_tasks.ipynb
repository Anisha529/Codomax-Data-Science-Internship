```python
#Saving Charts as Images
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data.csv')

#Bar Chart
x=df['Name']
y=df['Marks']
plt.figure(figsize=(6,4))
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Bar Chart of Student Marks',fontdict={'fontname':'Arial','fontsize':16,'fontweight':'bold'})
plt.bar(x,y)
plt.savefig('Charts/bar_chart.png')
plt.show()
plt.close()

#Line Chart
plt.figure(figsize=(10,5))
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Line Chart for Student Marks',fontdict={'fontname':'Comic Sans MS','fontsize':16,'fontweight':'bold'})
plt.plot(x,y)
plt.savefig('Charts/line_chart.png')
plt.show()
plt.close()

#Pie Chart
plt.figure(figsize=(5,5))
plt.title('Pie Chart for Student Marks',fontdict={'fontname':'Arial','fontsize':16,'fontweight':'bold'})
plt.pie(y,labels=x,autopct='%0.01f%%')
plt.savefig('Charts/pie_chart.png')
plt.show()
plt.close()
```
