```python
import pandas as pd
dict5={
    "Name":["Ankush","Preeti","Ayush","Banu","Kavya","Avinash"],
    "Marks":[88,94,84,76,98,83],
}
df=pd.DataFrame(dict5)
df.to_csv('data.csv',index=False)
print(df)
```

```python
#Data Analysis
import pandas as pd
df=pd.read_csv('data.csv')
df1=df['Marks'].sum()
df2=df['Marks'].mean()
df3=df['Marks'].max()
df4=df['Marks'].min()
df5=df['Marks'].count()

print("Sum of Marks:",df1)
print("Average Marks:",df2)
print("Maximum Marks:",df3)
print("Minimum Marks:",df4)
print("Number of Students:",df5)
```

```python
#Implementing Different Charts
#Bar Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data.csv')
x=df['Name']
y=df['Marks']
plt.figure(figsize=(10,5))
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Bar Chart of Student Marks')
plt.bar(x,y)
plt.show()
```

```python
#Implementing Different Charts
#Line Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data.csv')
x=df['Name']
y=df['Marks']
plt.figure(figsize=(10,5))
plt.xlabel('Name')
plt.ylabel('Marks')
plt.title('Line Chart for Student Marks')
plt.plot(x,y)
plt.show()
```

```python
#Implementing Different Charts
#Pie Chart
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('data.csv')
x=df['Name']
y=df['Marks']
plt.figure(figsize=(10,5))
plt.title('Pie Chart for Student Marks')
plt.pie(y,labels=x,autopct='%0.01f%%')
plt.show()
```
