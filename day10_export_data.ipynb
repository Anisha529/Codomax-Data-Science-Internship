```python
#Creating a messy dataset
import pandas as pd
import numpy as np
data={
    "Name":["Ankush","Preeti","Ayush","Banu","Kavya","Avinash"],
    "Age":[np.nan,21,23,np.nan,21,20],
    "Department":["CSE","ECE",np.nan ,"CSE","IT","IT"],
    "Marks":[88,94,84,76,98,83],
    "City":["Delhi",np.nan,"Kolkata","Lucknow",np.nan,"Hyderabad"]
}
df=pd.DataFrame(data)
df.to_csv('data2.csv')
print(df)
```

```python
#Cleaning the dataset
import pandas as pd
df=pd.read_csv('data2.csv')
df=df.dropna(subset=["Name","Age"])
df=df.fillna({"Department":"CSE","City":"Chennai","Age":"22"})
print(df)
```

```python
#Exporting the cleaned dataset to a new CSV file
df.to_csv('cleaned_data.csv',index=False)
```
